```
import os
import shutil
from IPython.display import clear_output

#@title <h1><b><font color=DODGERBLUE>📡 𝗩𝗝 𝗔𝗨𝗧𝗢 𝗙𝗢𝗥𝗪𝗔𝗥𝗗 𝗕𝗢𝗧</font></b>
# @markdown <div><center><a href="https://github.com/cosmos475/aesthetic-project"><img height="100" src="https://opengraph.githubassets.com/1/cosmos475/aesthetic-project"></center></a></div>
# @markdown <br><center><h2><strong><font color=DODGERBLUE>MTProto Auto-Forward Bot</font></strong></h2></center>
# @markdown <center><i>Polling-based — no public URL needed on Colab.</i></center>

# ── Repo settings ──
GITHUB_URL = "https://github.com/cosmos475/aesthetic-project.git"  #@param {type:"string"}
BRANCH_NAME = "main"  #@param {type:"string"}

base_dir = './repo'

def clone_or_update_repo(repo_url, base_directory, branch_name):
    repo_name = os.path.basename(repo_url).replace('.git', '')
    project_dir = os.path.join(base_directory, repo_name)

    if os.path.exists(project_dir):
        print(f"Deleting existing repository at: {project_dir} ...")
        shutil.rmtree(project_dir)
        print("Deleted existing repository successfully!")

    print(f"Cloning repository from {repo_url} (branch: {branch_name})...")
    clone_cmd = f"git clone -b {branch_name} {repo_url} {project_dir}"
    os.system(clone_cmd)
    print("Repository cloned successfully!")

    return project_dir

project_dir = clone_or_update_repo(GITHUB_URL, base_dir, BRANCH_NAME)

try:
    print(f"Entering project directory: {os.path.basename(project_dir)}...")
    os.chdir(project_dir)
    print("Entered project directory successfully!")
except FileNotFoundError as e:
    print(f"Error: {e}")
    raise

clear_output()

# Install required dependencies
print("Installing required dependencies...")
!pip install -q -r requirements.txt
print("All requirements installed successfully!")

#@markdown ### <font color=DODGERBLUE>🔑 Environment Variables — Mandatory</font>

API_ID = ""  #@param {type:"string"}
os.environ['API_ID'] = API_ID

API_HASH = ""  #@param {type:"string"}
os.environ['API_HASH'] = API_HASH

BOT_TOKEN = ""  #@param {type:"string"}
os.environ['BOT_TOKEN'] = BOT_TOKEN

BOT_OWNER = ""  #@param {type:"string"}
os.environ['BOT_OWNER'] = BOT_OWNER

DATABASE_URI = ""  #@param {type:"string"}
os.environ['DATABASE_URI'] = DATABASE_URI

#@markdown ### <font color=DODGERBLUE>⚙️ Environment Variables — Optional (khaali chhod sakte ho)</font>

DATABASE_NAME = "vj-forward-bot"  #@param {type:"string"}
os.environ['DATABASE_NAME'] = DATABASE_NAME

BOT_SESSION = "vjbot"  #@param {type:"string"}
os.environ['BOT_SESSION'] = BOT_SESSION

# PING_URL intentionally left unset on Colab — there is no stable public
# URL here, so keepalive self-pinging is skipped automatically (PING_URL
# defaults to empty, and app.py's keep_alive() just exits early with a
# warning log when it's not set — no crash).

clear_output()

# Run the bot (this will block and stream logs live — this cell keeps the
# Colab session alive automatically while it runs. To stop the bot, press
# the ■ stop button on this cell.)
RUN_COMMAND = "python3 app.py"
print(f"Running command: {RUN_COMMAND} ...")
!{RUN_COMMAND}
print("Execution completed!")
```
